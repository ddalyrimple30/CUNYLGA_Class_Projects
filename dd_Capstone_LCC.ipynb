```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import numpy as np

# Upload the file if running in Colab
# from google.colab import files
# uploaded = files.upload()

# Load the data
df = pd.read_excel("CapstoneProject.xlsx", sheet_name="Sheet1")

# --- CLEANING ---

# Standardize column names
df.columns = df.columns.str.strip().str.replace(" ", "_").str.replace("-", "_")

# Define numeric columns and convert
numeric_cols = ['Tot_Dschrgs', 'Avg_Submtd_Cvrd_Chrg', 'Avg_Tot_Pymt_Amt', 'Avg_Mdcr_Pymt_Amt']
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')

# Drop rows with missing key values
df.dropna(subset=['Tot_Dschrgs', 'Avg_Mdcr_Pymt_Amt'], inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Precalculate total Medicare payments
df['Total_Medicare_Payments'] = df['Tot_Dschrgs'] * df['Avg_Mdcr_Pymt_Amt']

# --- ANALYSIS ---

# Total discharges and total Medicare payments
total_discharges = int(df['Tot_Dschrgs'].sum())
total_medicare_payments = df['Total_Medicare_Payments'].sum()

print(f"Total Discharges: {total_discharges:,}")
print(f"Total Medicare Payments: ${total_medicare_payments:,.2f}")

# Top 5 Providers by Medicare Payments
top_providers = (
df.groupby('Rndrng_Prvdr_Org_Name')['Total_Medicare_Payments']
.sum()
.reset_index()
.sort_values(by='Total_Medicare_Payments', ascending=False)
.head(5)
)
print("\nTop 5 Providers by Medicare Payments:")
print(top_providers)

# Top 5 DRGs by Total Discharges
top_drgs_discharges = (
df.groupby('DRG_Desc')['Tot_Dschrgs']
.sum()
.reset_index()
.sort_values(by='Tot_Dschrgs', ascending=False)
.head(5)
)
print("\nTop 5 DRGs by Discharges:")
print(top_drgs_discharges)

# Top 5 DRGs by Average Medicare Payment
top_drgs_payments = (
df.groupby('DRG_Desc')['Avg_Mdcr_Pymt_Amt']
.mean()
.reset_index()
.sort_values(by='Avg_Mdcr_Pymt_Amt', ascending=False)
.head(5)
)
print("\nTop 5 DRGs by Avg Medicare Payment:")
print(top_drgs_payments)

# Medicare Payments by State
state_payments = (
df.groupby('Rndrng_Prvdr_State_Abrvtn')['Total_Medicare_Payments']
.sum()
.reset_index()
.sort_values(by='Total_Medicare_Payments', ascending=False)
)
print("\nMedicare Payments by State:")
print(state_payments.head())
```
